3. >  Working with Numbers:
    * Types:
        * type(): function is basically showing the data type of the numbers.

        * int(): function converts compatible value into int value. Like, 5, -12, 10000
        
        * float(): converts compatible value into float value. like,   3.16, -0.5, 100.0 

        * complex(): complex(real,imaginary) n.b. add +,- before/after to get negative value. 

    * Math:
        * for addition: +
        * for substitution: -
        * for multiplying: *
        * To divide values: /
        * // Floor Divison: it divides two number and rounds down  

        * % remainder: The leftover part after the division and also used to check if a number is even (0) / odd (1)

        +  ** exponentiation: (base ** exponent) It raises a number to the power of another number. to the power function. like, 2{base} to the power 3{exponent} = 8  

    * Rounding:
        * abs(value): returns the absolute (non-negative) value of a number. useful for measuring distance or size.

        * round(): we round the messy numbers, to make results easier to read and work with. it round up to the nearest whole number up or down depending what's closer.

            * round(number, ndigits): rounds the number to the specified number of decimal places

        * Bankers Rounding: ties (like 0.5) go to the nearest even number.
            
        * math.ceil(): rounds a number up to celling.

        * math.floor(): rounds a number down to the floor.

        * math.trunc(): cuts off the decimal part and keeps the whole number (no rounding).

    * Random:

        * random.random(): returns a random float between 0.0 and 1.0

        * random.randint(sart,end): gets a random whole number from start to end (both included)

    * Validating:
        
        * is_integer(): class method: output: bool:  checks if a float has no decimal part (is a whole number)

        * isinstance(value,type): output: bool  checks if a value belongs to a certain data type.
