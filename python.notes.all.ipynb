# Data Extraction: Indexing & Slicing:


* "hello" it's a string. but behind the scene, Pyhton thinks this as a characters and has a position number(index). Two types:

    > 1. Positive Index: which will start from- left to right.
        > First Index number is 0, "hello" will be positive indexing like this-
           > h=0, e=1, l=2, l=3, o=4

    > 2. Negative Index: which will start from- right to left.
        > First Index number is -1 "hello will be negative indexing like this-
            > h=-1, e=-2, l=-3, l=-4, o=-5 
* Indexing = is a specific value of a character; like here for the character "h" index number are: 0 and -5. so, to do that, here
 >   to print "h" provide the index number in [0] or [-5]

* Slicing = is same but this time it is characters, multiple characters will be sliced. so to do that,
provide the index number, like this [start:end] but,
   * you have to ensure that end number should be +1 from what we expected to be printed. so to do that, here 

   > * to print "hel" provide the index number [0:3] or [-5:-2] the end index number is +1 from what we expected to be seen. 

* open-ended slicing = 
   + this is just to print everything after the start point. to do that specify the start point and [0:] don't add any end point.
   
   + if i leave the start index empty, Python starts from index 0 to the mentioned index like this [:4] .

* [start:end:step] = step is if we don't want to add every character in the output. optional. like if we want to add every second characters then [0:4:2] then here output will be "hl" e and second l will be removed.    



```python
# Indexes & Slicing
text = "Python"

# Extract the first character:
print(text[0]) 
print(text[-6]) # this are the same index number the "P" character.

# Extract the last character:
print(text[5])
print(text[-1]) # this are the same index the "n" character 

# Extract first four characters: like the year.
date = "2026-09-20"
print(date[0:4]) 
print(date[:4]) # Python gonna start form index 0

# Extract the month:
print(date[5:7])

# Extract the Day:
print(date[8:10]) 
print(date[8:]) # this is also right we are expecting everything after the end start index
```

# Data Cleansing: Clean / Remove White-Spaces
* strip(): strr: method type: if we don't specify which side, this will remove both sides spaces. 
         
       * n.b. It removes the tabs space and multiple spaces.
       * n.b. It only removes the spaces at the start and end, not in the middle.
       * It removes any characters (have to indicate it to remove) you want from the start and end - not just spaces 
    * lstrip(): str: method type: removes white space from the left side of the string.
    * rstrip(): str: method type:  removes white space from the right side of the string.


```python
# Whitespace Cleanup:
text1 = " Engineering "
print(text1.strip()) #removes spaces from both side of a string

print("\t")
text1 = " Engineering "
print(text1.lstrip()) #removes spaces from the left side of a string

print("\t")
text1 = " Engineering "
print(text1.rstrip()) #removes spaces from the right side of a string

print("\t")
text1 = "Data Engineering"
print(text1.strip()) #It isn't removes spaces from the middle of a string

print("\t")
text1 = "###Engineering###"
print(text1.strip("#")) # removes characters from the both side of a string but have to indicate the character
```

# String Functions: Validating:
Checking and validating the quality of our string. We have some expectations on our values and we would like to check this expectations

>* isalpha() method type: checks if the string has only letters.
> * isnumeric() method type: checks if the string has only numbers.

```python
# Validation:
# isalpha(): check if the country name contains only letters/alphabetics:
country = "Bangladesh"
print(country.isalpha())  # We are getting True that means the variable has only letter.

country1 = "Bangladesh1"
print(country1.isalpha())  # We are getting False because it has now a number.

country1 = "Bangladesh#"
print(country1.isalpha())  # We are getting False because it has now a special character.
```

```python
# Validation:
# isnumeric(): checks if the string has only numbers: 
phone1 = "01700211097"
print(phone1.isnumeric())  # We are getting True that means the variable has only numbers.

phone1 = "0170-0211097"
print(phone1.isnumeric())  # We are getting False because the variable has not only numbers but also special characters.

#special mention on validation: isnumeric():
dot_number = "3.1416"
print(dot_number.isnumeric())  # It's supposed to be an True as it is a float number. But, We are getting False because dot(.) also considered as a special character.

# By Using this isalpha() and isnumeric() we can prevent invalid and garbage data entering your system.
```

```python
# Measure distance: abs():
# useful for measuring distance or size, regardless of direction.
print(2 - 10)  # we are getting the negative.
print(abs(2 - 10))  

# Rounding Numbers: round():
# sometimes numbers are messy, we round them to make results easier to read and work with.
price = 35.547896745877963
print(round(price))  # ties (like 0.5) go to the nearest even number.

import math
price = 35.547896745877963
print(round(price, 3)) # round(number, ndigits) rounds the number to the secified number of decimal places
print(math.floor(price))
print(math.ceil(price))

# Without Rounding math.trucn(value): 
import math 
price = 35.547896745877963
print(math.trunc(price))  # cuts off the decimal part and keeps the whole number (no rounding)
# another way:
print(int(price))  # don't have to import math module, but it's cuts off the decimal part and keeps the whole number (no rounding)
```

# Number Functions: Validation:
* is_integer(): class method: output: bool:  checks if a float has no decimal part (is a whole number)

* isinstance(value,type): output: bool  checks if a value belongs to a certain data type.

```python
# Number Functions: Validation: is_integer():
# check if the numebers are truly whole floats with .0 might just be from file exports 
x = 7.0
print(x.is_integer())  # boolean type thats why: True because 7.0 means 7 which is a whole number. 

x = 4.2
print(x.is_integer())  # False because 4.2 is not 4 or a whole number it has decimal part
```

```python
# Number Functions: Validation: isinstance():
# checks if a value belongs to a certain data type
x = 70 # this is a integer
print(isinstance(x, int)) # I am asking python is my value an integer?
print(isinstance(x, float))  # # I am asking python is my value an float?

x = 70.365 #Now, this a float
print(isinstance(x, int)) # I am asking python is my value an integer?
print(isinstance(x, float))  # # I am asking python is my value an float?
```
